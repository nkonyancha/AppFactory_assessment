```python
#binary search
    
def binary_search(nums,target):
    ### CONSTRAINTS CHECK FIRST ###
    if len(nums)<1 or len(nums)> 10000:
        return "list length is out of bounds"
    if target< -10000 or target>10000:
        return "target value is out of bounds"
    if any(num < -10000 or num > 10000 for num in nums):
        return "list values are out of bounds"
    
    first=0 # basically my cursor. this variable identifies starting point of the search. helps cut the list in half
    last=len(nums)-1 # identifies the end point of the search
    while first<=last:
        mid=(first+last)//2  #  this is the mid point. this is the value used to check against the target value.
        if nums[mid]==target:
            return mid # returns the index of the target value if found
        elif nums[mid]<target:
            first=mid+1#  this moves the starting point up if mid was less than the target value
        else:
            last=mid-1#  this moves the ending point down if mid was greater than target value
    return -1 # returns -1 if target value is not in the list.

nums=[-1,0,1,2,4,6,8,9,10]

print(binary_search(nums,8))




    
```
